```python
import torch
```

```python
x = torch.tensor(10.0)
y = torch.tensor([[3.0]])

w = torch.rand(1, 1, requires_grad=True)
b = torch.rand(1, 1, requires_grad=True)

z = w * x + b
print(z)
```

```python
loss = torch.nn.MSELoss()
loss_value = loss(z, y)
print(loss_value)
```

```python
# 必须只有损失值才能调用backward()方法
loss_value.backward()

print(w.grad)
print(b.grad)
```

```python
x = torch.tensor(2.0, requires_grad=True)
y = x.detach()

print(x.requires_grad)
print(y.requires_grad)

z1 = x ** 2
z2 = y ** 2

print(z1)
print(z2)
```

```python
z1.sum().backward()
# z2.backward()
```

```python
x = torch.ones(2, 2, requires_grad=True)
y = x ** 2

print(x)
print(y)

U = y.detach()
print(U)

z = U * x
z.sum().backward()
print(x.grad)
```
