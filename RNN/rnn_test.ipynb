```python
import torch
import torch.nn as nn
```

```python
rnn = nn.RNN(input_size=8, hidden_size=16, num_layers=2)
```

```python
# 定义输入数据和初始隐状态
# input:[seq_len （时间）序列长度, batch_size 批量大小, input_size 特征维度（必须和创建的RNN中的input_size保持一致）]
input = torch.randn(3, 3, 8)
# hx:[num_layers, batch_size, hidden_size]（全部和上述定义过的保持一致）
hx = torch.randn(2, 3, 16)
```

```python
# 前向传播
output, hn = rnn(input, hx)
print(output)
print(hn)
print(output.shape)
print(hn.shape)
```
